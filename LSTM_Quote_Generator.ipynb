```python
import numpy as np
import string
import pandas as pd
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Recreate and fit the tokenizer since it is missing in the current session
df = pd.read_csv('qoute_dataset.csv')
quotes = df['quote'].str.lower()
translator = str.maketrans('', '', string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))

vocab_size = 8978
tokenizer = Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)

text = "life is"
text = text.lower()
text = text.translate(str.maketrans('', '', string.punctuation))

sequence = tokenizer.texts_to_sequences([text])[0]

padded = pad_sequences(
    [sequence],
    maxlen=model.input_shape[1],
    padding='pre'
)

prediction = model.predict(padded, verbose=0)

predicted_id = np.argmax(prediction[0])

predicted_word = tokenizer.index_word.get(
    predicted_id,
    "Unknown"
)

print("Input:", text)
print("Predicted word:", predicted_word)
print("Predicted token ID:", predicted_id)
```

```python
def generate_text(seed_text, next_words=10):
    text = seed_text.lower()

    for _ in range(next_words):
        cleaned = text.translate(
            str.maketrans('', '', string.punctuation)
        )

        sequence = tokenizer.texts_to_sequences([cleaned])[0]

        padded = pad_sequences(
            [sequence],
            maxlen=model.input_shape[1],
            padding='pre'
        )

        prediction = model.predict(padded, verbose=0)
        predicted_id = np.argmax(prediction[0])

        if predicted_id == 0:
            break

        predicted_word = tokenizer.index_word.get(predicted_id)

        if predicted_word is None:
            break

        text += " " + predicted_word

    return text

print(generate_text("life is", next_words=10))
```
